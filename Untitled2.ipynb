```python
import os
import xml.etree.ElementTree as ET

def create_classes_file(classes_file, class_names):
    """
    Create a classes.txt file if it doesn't exist.
    """
    if not os.path.exists(classes_file):
        with open(classes_file, 'w') as f:
            f.write("\n".join(class_names) + "\n")
        print(f"Created classes file: {classes_file}")

def convert_voc_to_yolo(voc_dir, yolo_dir, classes_file):
    """
    Convert PascalVOC annotations to YOLO format.
    """
    os.makedirs(yolo_dir, exist_ok=True)


    default_classes = ["person", "hard-hat", "gloves", "mask", "glasses", "boots", "vest", "ppe-suit", "ear-protector", "safety-harness"]
    create_classes_file(classes_file, default_classes)


    with open(classes_file, 'r') as f:
        classes = [line.strip() for line in f.readlines()]

    if not os.path.exists(voc_dir):
        print(f"Error: VOC annotations directory '{voc_dir}' not found.")
        return

    annotation_files = [f for f in os.listdir(voc_dir) if f.endswith(".xml")]
    if not annotation_files:
        print("Error: No XML annotation files found in the specified directory.")
        return

    for xml_file in annotation_files:
        tree = ET.parse(os.path.join(voc_dir, xml_file))
        root = tree.getroot()

        img_width = int(root.find("size/width").text)
        img_height = int(root.find("size/height").text)

        yolo_annotations = []
        for obj in root.findall("object"):
            class_name = obj.find("name").text
            if class_name not in classes:
                continue
            class_id = classes.index(class_name)

            bndbox = obj.find("bndbox")
            xmin = int(bndbox.find("xmin").text)
            ymin = int(bndbox.find("ymin").text)
            xmax = int(bndbox.find("xmax").text)
            ymax = int(bndbox.find("ymax").text)


            x_center = (xmin + xmax) / (2.0 * img_width)
            y_center = (ymin + ymax) / (2.0 * img_height)
            width = (xmax - xmin) / img_width
            height = (ymax - ymin) / img_height

            yolo_annotations.append(f"{class_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}")


        yolo_filename = os.path.splitext(xml_file)[0] + ".txt"
        with open(os.path.join(yolo_dir, yolo_filename), 'w') as f:
            f.write("\n".join(yolo_annotations) + "\n")

    print(f"Successfully converted {len(annotation_files)} annotation files to YOLO format.")

voc_annotations_dir = "/mnt/data/datasets/labels"
yolo_annotations_dir = "/mnt/data/datasets/yolo_labels"
classes_file = "/mnt/data/datasets/classes.txt"
convert_voc_to_yolo(voc_annotations_dir, yolo_annotations_dir, classes_file)
```
